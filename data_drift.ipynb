```python
# Définir le chemin des fichiers (à adapter selon ton organisation Drive)
train_path = "/content/drive/MyDrive/Colab Notebooks/DS_projet_7/data_modelisation/app_train.csv"
test_path = "/content/drive/MyDrive/Colab Notebooks/DS_projet_7/data_modelisation/app_test.csv"
```

```python
# Charger les fichiers CSV
train_data = pd.read_csv(train_path)
test_data = pd.read_csv(test_path)
```

```python
# Vérification des données
print("Train Data Shape:", train_data.shape)
print("Test Data Shape:", test_data.shape)
```

- train.csv (application_train) représente les données utilisées pour entraîner le modèle.
- test.csv (application_test) représente les nouvelles données clients une fois le modèle en production

## Préparation des données

```python
from sklearn.impute import SimpleImputer

# Vérifier et enlever 'TARGET' si présent
if "TARGET" in train_data.columns:
    train_data = train_data.drop(columns=["TARGET"])
if "TARGET" in test_data.columns:
    test_data = test_data.drop(columns=["TARGET"])

train_data = train_data.drop(columns=['Unnamed: 0'], errors='ignore')
test_data = test_data.drop(columns=['Unnamed: 0'], errors='ignore')

# Vérifier les valeurs manquantes et appliquer le même traitement que pour la modélisation
imputer = SimpleImputer(strategy='median')

# Ajuster l'imputation sur `train_data` et appliquer aux deux datasets
train_data = pd.DataFrame(imputer.fit_transform(train_data), columns=train_data.columns)
test_data = pd.DataFrame(imputer.transform(test_data), columns=test_data.columns)

# Vérification après traitement
print("Train Data Shape:", train_data.shape)
print("Test Data Shape:", test_data.shape)
print("Nombre de valeurs manquantes après imputation dans train :", train_data.isnull().sum().sum())
print("Nombre de valeurs manquantes après imputation dans test :", test_data.isnull().sum().sum())
```

```python
print("Colonnes constantes dans train_data :")
print(train_data.nunique()[train_data.nunique() == 1])

print("Colonnes constantes dans test_data :")
print(test_data.nunique()[test_data.nunique() == 1])
```

```python
# Supprimer les colonnes constantes de test_data
cols_constantes = test_data.nunique()[test_data.nunique() == 1].index
test_data = test_data.drop(columns=cols_constantes)

print(f"Colonnes supprimées car constantes dans test_data : {list(cols_constantes)}")
print(f"Nouvelle shape de test_data : {test_data.shape}")
```

```python
# Supprimer les mêmes colonnes de train_data pour garantir la cohérence
train_data = train_data.drop(columns=cols_constantes, errors='ignore')

print(f"Colonnes supprimées dans train_data et test_data : {list(cols_constantes)}")
print(f"Nouvelle shape de train_data : {train_data.shape}")
print(f"Nouvelle shape de test_data : {test_data.shape}")
```

## Rapport

```python
from evidently.metric_preset import DataDriftPreset
from evidently.report import Report

# Le dataset de référence est `train.csv`, celui en production est `test.csv`
reference_data = train_data
current_data = test_data

# Création du rapport Evidently
data_drift_report = Report(metrics=[DataDriftPreset()])
data_drift_report.run(reference_data=reference_data, current_data=current_data)

# Sauvegarde et affichage du rapport
data_drift_report.save_html("data_drift_report.html")

from IPython.core.display import display, HTML
display(HTML("data_drift_report.html"))
```
